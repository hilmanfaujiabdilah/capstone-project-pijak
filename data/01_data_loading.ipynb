```python
# pip install pandas numpy matplotlib seaborn scikit-learn jupyter ipykernel openpyxl
# SELECT KERNEL : Python 3 3.12.7
# ============================================
# CELL 1: IMPORT LIBRARIES
# ============================================
import pandas as pd
import numpy as np
import os
from pathlib import Path

# Pengaturan tampilan
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

print("✅ Libraries imported successfully!")
print(f"📦 Pandas version: {pd.__version__}")
print(f"📦 Numpy version: {np.__version__}")
```

```python
# ============================================
# CELL 2: DEFINE PATH
# ============================================
# Path ke folder data/raw
DATA_PATH = Path(r'D:\CAPSTONE\data\raw')

# Cek folder ada
print(f"📁 Data path: {DATA_PATH}")
print(f"📁 Path exists: {DATA_PATH.exists()}")

# List semua CSV yang ada
csv_files = list(DATA_PATH.glob('*.csv'))
print(f"\n📄 Ditemukan {len(csv_files)} file CSV:")
for f in sorted(csv_files):
    size_mb = f.stat().st_size / (1024 * 1024)
    print(f"   ✅ {f.name:<45} ({size_mb:.2f} MB)")
```

```python
# ============================================
# CELL 3: LOAD SEMUA DATASET
# ============================================
print("🔄 Loading datasets...\n")

# 5 tabel utama + 1 translation
orders = pd.read_csv(DATA_PATH / 'olist_orders_dataset.csv')
print(f"✅ orders          : {orders.shape[0]:>7,} rows × {orders.shape[1]} cols")

order_items = pd.read_csv(DATA_PATH / 'olist_order_items_dataset.csv')
print(f"✅ order_items      : {order_items.shape[0]:>7,} rows × {order_items.shape[1]} cols")

products = pd.read_csv(DATA_PATH / 'olist_products_dataset.csv')
print(f"✅ products         : {products.shape[0]:>7,} rows × {products.shape[1]} cols")

customers = pd.read_csv(DATA_PATH / 'olist_customers_dataset.csv')
print(f"✅ customers        : {customers.shape[0]:>7,} rows × {customers.shape[1]} cols")

reviews = pd.read_csv(DATA_PATH / 'olist_order_reviews_dataset.csv')
print(f"✅ reviews          : {reviews.shape[0]:>7,} rows × {reviews.shape[1]} cols")

category_translation = pd.read_csv(DATA_PATH / 'product_category_name_translation.csv')
print(f"✅ category_transl  : {category_translation.shape[0]:>7,} rows × {category_translation.shape[1]} cols")

print("\n🎉 Semua dataset berhasil di-load!")
```

```python
# ============================================
# CELL 4: EKSPLORASI TABEL ORDERS
# ============================================
print("="*60)
print("📊 TABEL: ORDERS")
print("="*60)

print("\n🔸 Kolom & Tipe Data:")
print(orders.dtypes)

print(f"\n🔸 Preview data (3 baris pertama):")
orders.head(3)
```

```python
# Lanjutan Cell 4 - Missing Values
print("\n🔸 Missing Values di tabel orders:")
missing = orders.isnull().sum()
missing_pct = (missing / len(orders) * 100).round(2)

missing_df = pd.DataFrame({
    'jumlah_missing': missing,
    'persentase (%)': missing_pct
})

# Tampilkan hanya yang ada missing
ada_missing = missing_df[missing_df['jumlah_missing'] > 0]
if len(ada_missing) > 0:
    print(ada_missing)
else:
    print("   Tidak ada missing values ✅")

print(f"\n🔸 Status order yang ada:")
print(orders['order_status'].value_counts())
```

```python
# ============================================
# CELL 5: EKSPLORASI TABEL REVIEWS
# ============================================
print("="*60)
print("📊 TABEL: REVIEWS")
print("="*60)

print("\n🔸 Missing Values:")
missing_rev = reviews.isnull().sum()
missing_pct_rev = (missing_rev / len(reviews) * 100).round(2)

pd.DataFrame({
    'jumlah_missing': missing_rev,
    'persentase (%)': missing_pct_rev
})
```

```python
# Distribusi review score
print("\n🔸 Distribusi Review Score:")
score_dist = reviews['review_score'].value_counts().sort_index()
for score, count in score_dist.items():
    bar = '█' * (count // 2000)
    pct = count / len(reviews) * 100
    print(f"   Score {score}: {count:>6,} ({pct:.1f}%) {bar}")

print(f"\n🔸 Rata-rata review score: {reviews['review_score'].mean():.2f}")

# Cek berapa review yang punya teks komentar
punya_komentar = reviews['review_comment_message'].notna().sum()
print(f"\n🔸 Review dengan teks komentar: {punya_komentar:,} dari {len(reviews):,}")
print(f"   ({punya_komentar/len(reviews)*100:.1f}% punya teks — ini untuk Sentiment Analysis)")
```

```python
# ============================================
# CELL 6: RINGKASAN SEMUA TABEL
# ============================================

datasets = {
    'orders': orders,
    'order_items': order_items,
    'products': products,
    'customers': customers,
    'reviews': reviews,
    'category_translation': category_translation
}

print("="*70)
print("📋 RINGKASAN SEMUA DATASET")
print("="*70)
print(f"{'Nama':<25} {'Rows':>10} {'Cols':>6} {'Missing Values':>15} {'Duplikat':>10}")
print("-"*70)

for name, df in datasets.items():
    rows = df.shape[0]
    cols = df.shape[1]
    missing = df.isnull().sum().sum()
    duplicates = df.duplicated().sum()
    print(f"{name:<25} {rows:>10,} {cols:>6} {missing:>15,} {duplicates:>10,}")

print("="*70)
print("\n⚠️  Kolom dengan missing values = perlu ditangani di Data Cleaning")
print("⚠️  Duplikat > 0 = perlu dihapus di Data Cleaning")
```

```python
# ============================================
# CELL 7: SIMPAN CATATAN EKSPLORASI
# ============================================

# Simpan info dasar ke processed folder
# (Berguna untuk dokumentasi & README)

summary_lines = []
summary_lines.append("# Data Exploration Summary\n")
summary_lines.append(f"Tanggal eksplorasi: {pd.Timestamp.now().strftime('%Y-%m-%d %H:%M')}\n\n")

for name, df in datasets.items():
    summary_lines.append(f"## {name}\n")
    summary_lines.append(f"- Shape: {df.shape[0]:,} rows × {df.shape[1]} cols\n")
    summary_lines.append(f"- Missing values: {df.isnull().sum().sum():,}\n")
    summary_lines.append(f"- Duplicates: {df.duplicated().sum():,}\n")
    summary_lines.append(f"- Columns: {', '.join(df.columns.tolist())}\n\n")

# Simpan ke file
output_path = Path(r'D:\CAPSTONE\data\processed\exploration_summary.md')
with open(output_path, 'w') as f:
    f.writelines(summary_lines)

print(f"✅ Summary tersimpan di: {output_path}")
print("\n🎉 Notebook 01 selesai!")
print("➡️  Langkah berikutnya: 02_data_cleaning.ipynb")
```
